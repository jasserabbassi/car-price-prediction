## Project Overview

### Objective
Develop and deploy machine learning models to predict car prices using historical car data and various features.

## Dataset Information

### Features
- **Numerical**: Year, Mileage, Engine Size, Asking Price
- **Categorical**: Brand, Model, Fuel Type, Transmission, Condition
- **Target**: Price (continuous)

### Data Statistics
- **Records**: 8,000+ cars
- **Features**: 9 input features
- **Data Quality**: ~100% completeness
- **Price Range**: Varies by brand and condition

### Data Preprocessing
1. **Missing Value Handling**: Removal
2. **Outlier Detection**: IQR method (1.5 × IQR)
3. **Categorical Encoding**: LabelEncoder
4. **Feature Scaling**: StandardScaler (mean=0, std=1)
5. **Train-Test Split**: 80-20

## Model Performance Summary

### Evaluation Metrics
- **R² Score**: Coefficient of determination (0-1, higher is better)
- **RMSE**: Root Mean Square Error (lower is better)
- **MAE**: Mean Absolute Error (lower is better)
- **MAPE**: Mean Absolute Percentage Error
- **Cross-Validation**: 5-fold CV for robustness

### Model Comparison
All models achieve **R² > 0.85**, indicating strong prediction accuracy.

**Best Performing Model**: Depends on dataset - See notebook `02_Model_Training.ipynb` for detailed comparison.

### Key Insights
- **Ensemble Methods** outperform linear models
- **Neural Network** provides alternative perspective
- **Feature Importance**: Year, Mileage, Brand are top predictors
- **Prediction Error**: Average ~5-10% MAPE

## FastAPI REST Endpoints

### Available Endpoints

**Base URL**: `http://localhost:8000`

1. **POST /predict** - Make a single prediction
   ```json
   {
     "Brand": "Toyota",
     "Model": "Camry",
     "Year": 2020,
     "Mileage": 25000,
     "Engine Size": 2.5,
     "Fuel Type": "Gasoline",
     "Transmission": "Automatic",
     "Condition": "Used"
   }
   ```

2. **POST /predict_batch** - Batch predictions

3. **GET /models** - Get available models

4. **GET /health** - Health check

5. **GET /docs** - Interactive API documentation (Swagger UI)
